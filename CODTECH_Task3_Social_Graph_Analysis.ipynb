```python

import networkx as nx
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python

G = nx.karate_club_graph()

plt.figure(figsize=(10, 7))
pos = nx.spring_layout(G, seed=42)

nx.draw(
    G, pos,
    with_labels=True,
    node_color="skyblue",
    node_size=700,
    edge_color="gray",
    font_size=9
)

plt.title("Social Network Graph Analysis")
plt.show()

print("Number of users:", G.number_of_nodes())
print("Number of connections:", G.number_of_edges())
```

```python

centrality = nx.degree_centrality(G)

print("Top 5 most connected users:")

for user, score in sorted(
    centrality.items(),
    key=lambda item: item[1],
    reverse=True
)[:5]:
    print(f"User {user}: {score:.3f}")
```

```python

print("Network Analysis Results")
print("Number of users:", G.number_of_nodes())
print("Number of connections:", G.number_of_edges())
print("Average degree:", round(
    sum(dict(G.degree()).values()) / G.number_of_nodes(), 2
))
print("Network density:", round(nx.density(G), 3))
print("Number of connected components:",
      nx.number_connected_components(G))
```
